Notes: 
- Sequence Modelling: When we say sequence modelling, we will be thinking of text. Sequences could also be:
    - Time series
    - DNA nucleotides
    - Images
- If we compare 3 models that have the same size(memory):
    - 20 params each 32 bits
    - 40 params each 16 bits
    - 80 params each 8 bits
- It turns out that **more params** that are not as accurate **are better** than **less params** that are more accurate (floating points) so **80 by 8 bits > 20 by 32 bits**
- More params: more complexity. This way we get more complexity for the same memory
- [4-bits per parameter seems to be optimal - Study](https://arxiv.org/pdf/2212.09720)
    - Conclusions from study:
    - <img src="images/4-bit-precision-graph.png" width=400/>
- **Gemini** explanation: Why is this true? (The Logic)
    - Information Density: A 32-bit float is extremely precise, but most neural network weights don't actually need that much "room." Much of those 32 bits is "wasted" on precision that doesn't change the final output.

    - Capacity vs. Precision: Increasing the number of parameters (the "80 params" in your teacher's example) adds more "dimensions" or "neurons" to the model. Having more neurons that are slightly "blurry" (lower bits) is better for reasoning than having very few neurons that are perfectly sharp.

    - The "Sweet Spot": The research found that 4-bit is the magic number. If you go down to 2-bit or 1-bit, the "blurriness" (quantization error) becomes so high that the model's intelligence collapses. 4-bit provides the maximum "intelligence per bit."

## [Standard ML algorithms vs Neural Networks](https://aws.amazon.com/compare/the-difference-between-machine-learning-and-deep-learning/#:~:text=ML%20is%20best%20for%20well,make%20sense%20of%20unstructured%20data.&text=ML%20solves%20problems%20through%20statistics,mathematics%20with%20neural%20network%20architecture.)
[More](https://www.geeksforgeeks.org/difference-between-artificial-intelligence-vs-machine-learning-vs-deep-learning/)


Standard ML:
- Pros:
    - If you have a clear idea about the logic(math) involved in behind and you can visualize the complex functionalities like K-Mean, Support Vector Machines, etc., then it defines the ML aspect.
    - **Work well with tabulated data**
    - ML is less complex and has a lower data volume.
- Cons:
    - Less efficient than DL as it can’t work for longer dimensions or higher amount of data.
    - Doesn't work well with unstructured data.
    - You have to manually select and extract features from raw data and assign weights to train an ML model.
     
NNs:
- Pros:
    - **The freedome they give us**
    - **We can create our own loss function (that is differentiable), hence we can solve any problem that we can define a loss function for**
    - More powerful than ML as it can easily work for larger sets of data.
    - **Works great with unstructured data (untabulated data).**
    - **Utilize GPUs better, but VRAM is often a bottleneck**
    - **Can work with huge amounts of parameters (billions) $10^{11}$**
    - Unlimited capacity
    - No feature selection/**creation** (A NN can figure out what a nose/face is on its own) (Deep learning models can self-learn using feedback from known errors.)
- Cons:
    - If you are clear about the math involved in it but don’t have idea about the features, so you break the complex functionalities into linear/lower dimension features by adding more layers, then it defines the DL aspect.
    - **Can't interpret the weights. Black box.**
    - Doesn't work well with tabulated data
    - Deep learning is more complex with a very high data volume.
 

Notes:
- All of the prior knowledge of ML is valid for DL. DL is a subset of ML, so ML has some additional things (like statistical models etc.) but it is the same thing. They are not 2 different things.
    - In DL we can do cross-validation as well. Also, we don't always do it in ML as well.
    - In DL we will see things that expand on the ML concepts.
- DNN are not meant for tabular data.
    - Tabular data has irregular patterns that could be easily split by a tree ensable model ( X > 50 ) where as DNN are using gradients adn make smooth curves which often leads to loss of accuracy.
    - Tabular data is usually in the range of 10 000 to 100 000, which is very little. DNNs require millions of samples. 
- In a tabular dataset, when adding features, all we are doing is help the model see the data better. The information stays the same, we are just adding better view points. (log(X) instead of just X column, we add a bit of bias this way, we are forcing the model to look at the data through log())
- **Ablation**: is a technique where you systematically remove components (like layers, features, or regularizers) from a neural network to understand each part's specific contribution to the model's overall performance, helping to identify crucial elements, simplify models, and improve explainability

## [CUDA](https://developer.nvidia.com/cuda-toolkit) (Compute Unified Device Architecture)
GPUs are made to render object on the screen, but we don't want to use them in this way. To change that we use CUDA.

Cuda is a toolkit that allows the GPU to make general purpose computing tasks instead of render objects on the screen.

## [cuDNN](https://developer.nvidia.com/cudnn)

cuDNN - CUDA® Deep Neural Network

cuDNN is a GPU-accelerated library of primitives(lego building blocks) for deep neural networks. It is build on top of CUDA.
cuDNN provides highly tuned implementations for standard routines such as forward and backward convolution, attention, matmul, pooling, and normalization.
This way Pytorch and Tensorflow don't have to implement these on their own for NVIDIA GPUs. They just use these primities and build abstractions on top of them.

## What is a [tensor](https://youtu.be/L35fFDpwIM4?t=504)?

Tensor = Multidimensional matrix ( [is an algebraic object that describes a multilinear relationship between sets of algebraic objects related to a vector space](https://en.wikipedia.org/wiki/Tensor) )

Tensors don't just hold the data in a Vector/Matrix/Multidimentional Array form. [**They are designed in a way to allow easy math calculations by GPUs on the data**](https://youtu.be/L35fFDpwIM4?t=413). Tensors also take care of the chain rule derivative calculations.

- Scalar = Tensor with rank 0
- Vector = Tensor with rank 1
- Matrix = Tensor with rank 2
- etc.

Example:
Image is a tensor with 3 ranks: Width, Height and 3 channels

#### Optimizer
During training, your model calculates a gradient (the direction of the steepest "hill"). The optimizer uses that gradient to adjust the model's weights to minimize the loss. Mathematically, it's solving an optimization problem where we want to find the minimum of the error surface.

The optimizer is the implementation of the gradient descent algorithm. If you think of Gradient Descent as the "strategy," the Optimizer is the "engine" that actually executes the math.

To be precise, the optimizer performs backpropagation. It calculates how much each weight in your network contributed to the error and then nudges those weights in the opposite direction of the gradient to reduce the loss.

We can see that the output is not normalized to be between 0 and 1, so we have to add softmax to the last Dense layer: `Dense(n_classes, activation = "softmax"),`


````python
neural_net = Sequential([
    Input(shape = input_shape),
    Flatten(),
    # to be Deep nn we can put as many layers we want in here
    Dense(n_classes, activation = "softmax"),
])
neural_net.summary()

neural_net.compile(loss = "sparse_categorical_crossentropy", optimizer = "adam")

neural_net.fit(attributes_train, labels_train, batch_size = 16, epochs = 10)
````

````python
neural_net(attributes_train[:5])
````

Much better predictions

Most importantly the sum of every row is 1, which means now we actually have probabilities

We see the `.fit()` did 1875 steps. The dataset is 60k, there fore it is not stochastic it is not batch GD it is mini-batch gradient descend. 32 rows per batch by default, can be changed.

How to determine batch size:
- needs to fit + the model weights in the video card ram
- needs to have atleast 1 batch to train with
- better to have atleast 3/4 to utilize loading/training etc processes. Utilize time better. The train can be faster than the load.
- Big batch size means the train would be stable
- Small batch size means the train would allow for bigger model to fit in the RAM
- Usually use a power of 2 (2,4,8 ...)

#### Epoch

One pass through the dataset is called an epoch. We need usually thousands of them or even more.

If we interrupt the model training, nothingwould be lost. Each epoch is overwriting the weights therefore it is saved.

We also have metrics. In this case we can track accuracy.

````python
keras.backend.clear_session()
neural_net = Sequential([
    Input(shape = input_shape),
    Flatten(),
    # to be Deep nn we can put as many layers we want in here
    Dense(n_classes, activation = "softmax"),
])
neural_net.summary()

neural_net.compile(loss = "sparse_categorical_crossentropy", optimizer = "adam", metrics = ["accuracy"])

neural_net.fit(attributes_train, labels_train, batch_size = 16, epochs = 10)
````

````python
neural_net.evaluate(attributes_train, labels_train)
````

### Convolutional Neural Networks (CNN)

#### Advice
If something doesn't work, first check the shapes of the layers.

````python
keras.backend.clear_session()
conv_net = Sequential([
    Input(shape = input_shape),
    # to be Deep nn we can put as many layers we want in here
    Conv2D(200, (3,3), activation = "relu"),
    Conv2D(100, (3,3), activation = "relu"),
    Conv2D(50, (3,3), activation = "relu"),
    Dense(n_classes, activation = "softmax"),
])
conv_net.summary()

conv_net.compile(loss = "sparse_categorical_crossentropy", optimizer = "adam", metrics = ["accuracy"])

conv_net.fit(attributes_train, labels_train, batch_size = 16, epochs = 10, validation_split=0.1)
````

````python
keras.backend.clear_session()
conv_net = Sequential([
    Input(shape = (28, 28, 1)),
    # to be Deep nn we can put as many layers we want in here
    Conv2D(200, (3,3), padding = "same", activation = "relu"),
    Conv2D(100, (3,3), padding = "same", activation = "relu"),
    Conv2D(50, (3,3), padding = "same", activation = "relu"),
    Flatten(),
    Dense(n_classes, activation = "softmax"),
])
conv_net.summary()

conv_net.compile(loss = "sparse_categorical_crossentropy", optimizer = "adam", metrics = ["accuracy"])

conv_net.fit(attributes_train, labels_train, batch_size = 16, epochs = 10, validation_split=0.1)
````

````python
keras.backend.clear_session()
conv_net = Sequential([
    Input(shape = (28, 28, 1)),
    # to be Deep nn we can put as many layers we want in here
    Conv2D(200, (3,3), activation = "relu"),
    MaxPool2D(),
    Conv2D(100, (3,3), activation = "relu"),
    MaxPool2D(),
    Conv2D(50, (3,3), activation = "relu"),
    Flatten(),
    Dense(n_classes, activation = "softmax"),
])
conv_net.summary()

conv_net.compile(loss = "sparse_categorical_crossentropy", optimizer = "adam", metrics = ["accuracy"])

conv_net.fit(attributes_train, labels_train, batch_size = 16, epochs = 5, validation_split=0.1)
````
