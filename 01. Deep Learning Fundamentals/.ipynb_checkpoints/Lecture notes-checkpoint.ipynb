## 1. Curriculum

Notes: 
- Sequence Modelling: When we say sequence modelling, we will be thinking of text. Sequences could also be:
    - Time series
    - DNA nucleotides
    - Images
- If we compare 3 models that have the same size(memory):
    - 20 params each 32 bits
    - 40 params each 16 bits
    - 80 params each 8 bits
- It turns out that **more params** that are not as accurate **are better** than **less params** that are more accurate (floating points) so **80 by 8 bits > 20 by 32 bits**
- More params: more complexity. This way we get more complexity for the same memory
- [4-bits per parameter seems to be optimal - Study](https://arxiv.org/pdf/2212.09720)
    - Conclusions from study:
    - <img src="images/4-bit-precision-graph.png" width=400/>
- **Gemini** explanation: Why is this true? (The Logic)
    - Information Density: A 32-bit float is extremely precise, but most neural network weights don't actually need that much "room." Much of those 32 bits is "wasted" on precision that doesn't change the final output.

    - Capacity vs. Precision: Increasing the number of parameters (the "80 params" in your teacher's example) adds more "dimensions" or "neurons" to the model. Having more neurons that are slightly "blurry" (lower bits) is better for reasoning than having very few neurons that are perfectly sharp.

    - The "Sweet Spot": The research found that 4-bit is the magic number. If you go down to 2-bit or 1-bit, the "blurriness" (quantization error) becomes so high that the model's intelligence collapses. 4-bit provides the maximum "intelligence per bit."
